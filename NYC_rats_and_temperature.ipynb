```python
import kagglehub

# Download latest version of the NYC Rat Sightings dataset
path = kagglehub.dataset_download("new-york-city/nyc-rat-sightings")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import glob
import os

# Find all CSV files in the downloaded folder
csv_files = glob.glob(os.path.join(path, "*.csv"))

if csv_files:
    print(f"Loading file: {csv_files[0]}")
    df = pd.read_csv(csv_files[0])
    display(df.head())
else:
    print("No CSV files found in the dataset folder. Listing all files:")
    print(os.listdir(path))
```

```python
# Configure pandas to display all columns without truncation
pd.set_option('display.max_columns', None)

# Display the first 5 rows with all columns visible
display(df.head())
```

```python
print(df.info())
## need to covert dates to datetime
# convert zip code to whole number
##drop all non-nulls, school columns
print(df['Complaint Type'].unique())
```

```python
# Convert date columns to datetime objects
df['Created Date'] = pd.to_datetime(df['Created Date'], errors='coerce')
df['Closed Date'] = pd.to_datetime(df['Closed Date'], errors='coerce')

# Display the updated types and first few rows
print(df[['Created Date', 'Closed Date']].dtypes)
display(df[['Created Date', 'Closed Date']].head())
```

```python
# Convert Incident Zip to nullable integer type
df['Incident Zip'] = df['Incident Zip'].astype('Int64')

# Display the unique values or a sample to verify
print("Data type:", df['Incident Zip'].dtype)
display(df[['Incident Zip']].head(10))
```

```python
# Identify columns that contain 'School' (case-insensitive)
school_cols = [col for col in df.columns if 'school' in col.lower()]
print(f"Dropping {len(school_cols)} columns: {school_cols}")

# Drop the columns
df.drop(columns=school_cols, inplace=True)

# Display remaining columns
print(f"Remaining columns: {len(df.columns)}")
display(df.head(2))
```

```python
# Calculate the percentage of NaN values for each column
nan_percentage = (df.isnull().sum() / len(df)) * 100

# Combine into a DataFrame for clean visualization
nan_df = pd.DataFrame({
    'Missing Values Count': df.isnull().sum(),
    'Percentage (%)': nan_percentage
}).sort_values(by='Percentage (%)', ascending=False)

# Display the results
display(nan_df)
```

```python
# Identify columns with 100% missing values
cols_to_drop = nan_percentage[nan_percentage == 100].index.tolist()
print(f"Dropping {len(cols_to_drop)} columns with 100% NaNs: {cols_to_drop}")

# Drop these columns from df
df.drop(columns=cols_to_drop, inplace=True)

# Print the remaining columns list and count
print(f"Remaining columns count: {len(df.columns)}")
display(df.head(2))
```

```python
# 1. Inspect rows where 'Incident Zip' is missing
missing_zip_df = df[df['Incident Zip'].isnull()]
print(f"Number of missing Incident Zips: {len(missing_zip_df)}")
display(missing_zip_df[['Incident Address', 'Street Name', 'City', 'Borough']].head(10))
```

```python
# Calculate count of sightings per borough
borough_counts = df['Borough'].value_counts()

# Calculate percentage of sightings per borough
borough_percents = df['Borough'].value_counts(normalize=True) * 100

# Combine into a summary DataFrame
borough_summary = pd.DataFrame({
    'Count': borough_counts,
    'Percentage (%)': borough_percents
})

display(borough_summary)
```

```python
# Filter and display the record with 'Unspecified' borough
unspecified_borough_df = df[df['Borough'] == 'Unspecified']
display(unspecified_borough_df)
```

```python
# Remove the record where Borough is 'Unspecified'
df = df[df['Borough'] != 'Unspecified']

# Confirm the record has been removed by checking the shape
print("New DataFrame shape:", df.shape)
print("Remaining Unspecified boroughs:", (df['Borough'] == 'Unspecified').sum())
```

```python
# Extract Year, Month, and Year-Month period for plotting
df['Year'] = df['Created Date'].dt.year
df['Month'] = df['Created Date'].dt.month
df['Year_Month'] = df['Created Date'].dt.to_period('M')

# Display a sample to confirm the new columns
display(df[['Created Date', 'Year', 'Month', 'Year_Month']].dropna().head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the plotting style
sns.set_theme(style="whitegrid")

# 1. Group overall sightings by Year-Month
overall_trend = df.groupby('Year_Month').size().reset_index(name='Sightings')
overall_trend['Year_Month_Str'] = overall_trend['Year_Month'].astype(str)

# Plot Overall Sightings over Time
plt.figure(figsize=(15, 6))
sns.lineplot(data=overall_trend, x='Year_Month_Str', y='Sightings', marker='o', color='crimson', linewidth=2)
plt.title('Overall Rat Sightings in NYC by Month/Year', fontsize=16)
plt.xlabel('Month/Year', fontsize=12)
plt.ylabel('Number of Sightings', fontsize=12)
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
import statsmodels.api as sm
import numpy as np

# 1. Prepare data for LOESS
# We will use the index of the overall_trend DataFrame as the numeric x-value
overall_trend = overall_trend.sort_values('Year_Month').reset_index(drop=True)
x_numeric = np.arange(len(overall_trend))
y_values = overall_trend['Sightings'].values

# Calculate lowess (LOESS) smoothing
# frac=0.2 means using 20% of the data points to estimate each local value
lowess_fit = sm.nonparametric.lowess(y_values, x_numeric, frac=0.2)
overall_trend['LOESS'] = lowess_fit[:, 1]

# 2. Plot Overall Sightings with LOESS Trend Line
plt.figure(figsize=(15, 6))
sns.lineplot(data=overall_trend, x='Year_Month_Str', y='Sightings', marker='o', color='crimson', alpha=0.4, label='Actual Sightings', linewidth=1)
sns.lineplot(data=overall_trend, x='Year_Month_Str', y='LOESS', color='darkblue', linewidth=3, label='LOESS Trend Line')

plt.title('Overall Rat Sightings in NYC with LOESS Trend Line', fontsize=16)
plt.xlabel('Month/Year', fontsize=12)
plt.ylabel('Number of Sightings', fontsize=12)
plt.xticks(rotation=90)
plt.legend(fontsize=12)
plt.tight_layout()
plt.show()
```

```python
from statsmodels.tsa.seasonal import seasonal_decompose

# Set Year_Month as the index and sort for time series analysis
trend_ts = overall_trend.set_index(pd.PeriodIndex(overall_trend['Year_Month'], freq='M'))['Sightings']

# Perform seasonal decomposition (additive model is appropriate since the amplitude of seasonal fluctuations is relatively stable)
decomposition = seasonal_decompose(trend_ts.to_timestamp(), model='additive')

# Plot the decomposed components
fig, (ax1, ax2, ax3, ax4) = plt.subplots(4, 1, figsize=(15, 12), sharex=True)

decomposition.observed.plot(ax=ax1, color='crimson')
ax1.set_title('Observed (Actual Sightings)', fontsize=12)
ax1.set_ylabel('Sightings')

decomposition.trend.plot(ax=ax2, color='darkblue', linewidth=2)
ax2.set_title('Isolated Trend Component', fontsize=12)
ax2.set_ylabel('Trend')

decomposition.seasonal.plot(ax=ax3, color='forestgreen')
ax3.set_title('Seasonal Component', fontsize=12)
ax3.set_ylabel('Seasonal')

decomposition.resid.plot(ax=ax4, style='o', color='gray', alpha=0.6)
ax4.set_title('Residuals (Noise)', fontsize=12)
ax4.set_ylabel('Residuals')

plt.suptitle('Seasonal Decomposition of NYC Rat Sightings (Monthly)', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# 2. Group sightings by Year-Month and Borough
borough_trend = df.groupby(['Year_Month', 'Borough']).size().reset_index(name='Sightings')
borough_trend['Year_Month_Str'] = borough_trend['Year_Month'].astype(str)

# Plot Sightings over Time broken down by Borough
plt.figure(figsize=(15, 8))
sns.lineplot(data=borough_trend, x='Year_Month_Str', y='Sightings', hue='Borough', marker='o', linewidth=2)
plt.title('NYC Rat Sightings by Month/Year and Borough', fontsize=16)
plt.xlabel('Month/Year', fontsize=12)
plt.ylabel('Number of Sightings', fontsize=12)
plt.xticks(rotation=90)
plt.legend(title='Borough', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns

# Get the list of unique boroughs
boroughs = sorted(df['Borough'].unique())

# Set up the subplots grid (e.g., 3 rows, 2 columns to fit 5 boroughs nicely)
fig, axes = plt.subplots(3, 2, figsize=(16, 18), sharex=True)
axes = axes.flatten()  # Flatten the 2D array of axes for easy indexing

# Configure global plotting parameters
sns.set_theme(style="whitegrid")

for i, borough in enumerate(boroughs):
    ax = axes[i]

    # Filter and group data for the specific borough
    b_data = df[df['Borough'] == borough].groupby('Year_Month').size().reset_index(name='Sightings')
    b_data = b_data.sort_values('Year_Month').reset_index(drop=True)
    b_data['Year_Month_Str'] = b_data['Year_Month'].astype(str)

    # Skip if there's not enough data for LOESS
    if len(b_data) > 5:
        x_numeric = np.arange(len(b_data))
        y_values = b_data['Sightings'].values

        # Calculate lowess (LOESS) smoothing
        lowess_fit = sm.nonparametric.lowess(y_values, x_numeric, frac=0.2)
        b_data['LOESS'] = lowess_fit[:, 1]

        # Plot actual sightings
        sns.lineplot(data=b_data, x='Year_Month_Str', y='Sightings', marker='o', ax=ax,
                     color='crimson', alpha=0.3, label='Actual', linewidth=1)

        # Plot LOESS trend
        sns.lineplot(data=b_data, x='Year_Month_Str', y='LOESS', ax=ax,
                     color='darkblue', linewidth=2.5, label='LOESS Trend')
    else:
        # Simple lineplot if we don't have enough points
        sns.lineplot(data=b_data, x='Year_Month_Str', y='Sightings', marker='o', ax=ax,
                     color='crimson', label='Actual')

    ax.set_title(f'{borough} - Rat Sightings & LOESS Trend', fontsize=14)
    ax.set_ylabel('Sightings', fontsize=10)
    ax.set_xlabel('Month/Year', fontsize=10)
    ax.tick_params(axis='x', rotation=90, labelsize=8)
    ax.legend(fontsize=10)

    # Floor the y-axis to start at 0
    ax.set_ylim(bottom=0)

# Hide any unused axes (since we have 5 boroughs and 6 subplot slots)
if len(boroughs) < len(axes):
    for j in range(len(boroughs), len(axes)):
        fig.delaxes(axes[j])

plt.suptitle('Monthly Rat Sightings & LOESS Trends by Borough', fontsize=18, y=0.98)
plt.tight_layout()
plt.show()
```

```python
import statsmodels.api as sm

# Prepare the data
# We will use 'Temp_Mean_F' and a numeric time index 'Time_Trend' as predictors
regression_data = correlation_df.copy()
regression_data['Time_Trend'] = range(len(regression_data))

# Define independent variables (X) and dependent variable (y)
X = regression_data[['Temp_Mean_F', 'Time_Trend']]
y = regression_data['Sightings']

# Add a constant (intercept) to the predictors
X = sm.add_constant(X)

# Fit the OLS model
model = sm.OLS(y, X).fit()

# Print the model summary
print(model.summary())
```

```python
# Calculate fitted values and residuals
fitted_values = model.fittedvalues
residuals = model.resid

# Create a 1x2 subplot structure for residuals diagnostics
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# 1. Residuals vs Fitted plot (Checks homoscedasticity and linearity)
sns.scatterplot(x=fitted_values, y=residuals, color='crimson', alpha=0.8, ax=axes[0])
axes[0].axhline(y=0, color='darkblue', linestyle='--', linewidth=2)
axes[0].set_title('Residuals vs Fitted Values', fontsize=14)
axes[0].set_xlabel('Fitted Values (Predicted Sightings)', fontsize=12)
axes[0].set_ylabel('Residuals', fontsize=12)

# 2. Normal Q-Q Plot (Checks normality of residuals)
stats.probplot(residuals, dist="norm", plot=axes[1])
axes[1].get_lines()[0].set_color('crimson')
axes[1].get_lines()[0].set_alpha(0.8)
axes[1].get_lines()[1].set_color('darkblue')
axes[1].get_lines()[1].set_linewidth(2)
axes[1].set_title('Normal Q-Q Plot', fontsize=14)
axes[1].set_xlabel('Theoretical Quantiles', fontsize=12)
axes[1].set_ylabel('Ordered Values', fontsize=12)

plt.suptitle('OLS Regression Residuals Diagnostics', fontsize=16, y=0.98)
plt.tight_layout()
plt.show()
```

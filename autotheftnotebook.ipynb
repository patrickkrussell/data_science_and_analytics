```python
# -*- coding: utf-8 -*-
"""
Created on Thu Jul 30 19:10:10 2026

@author: russe
"""

import pandas as pd

auto_theft_data = pd.read_csv(r"C:/Users/russe/Downloads/Auto_Theft_Open_Data_4481082360476864088.csv")

df_info = auto_theft_data.info()
df_described = auto_theft_data.describe()



#convert Report and Occ dates to datetime
date_cols = ['REPORT_DATE', 'OCC_DATE']

auto_theft_data[date_cols] = auto_theft_data[date_cols].apply(pd.to_datetime)


#checking for na's
na_check = auto_theft_data.isna().any()

#empty values in OCC YEAR, MONTH, DAY, DOY, DOW
na_check_2 = auto_theft_data.isna().sum()

na_check_3_by_row = auto_theft_data[auto_theft_data.isna().any(axis=1)]



auto_theft_data['OCC_YEAR'].value_counts() #not much before 2014

#take out any OCC_date before 2014
auto_theft_data_cleaning = auto_theft_data[
    (auto_theft_data['REPORT_YEAR'] >= 2014) &
    (auto_theft_data['REPORT_YEAR'] < 2026)
] ##cut off 2026 too

#check for dupilicates
df_duplicates = auto_theft_data_cleaning[auto_theft_data_cleaning.duplicated()]
duplicated_id = auto_theft_data_cleaning[auto_theft_data_cleaning.duplicated(subset=['EVENT_UNIQUE_ID'], keep=False)]
#14098 duplicated event_unique_id - what does this mean?


#check yearly totals with TPS Dashboard to see if the dupliacted records are counted or not
overall = (
    auto_theft_data_cleaning
    .groupby("REPORT_YEAR")["EVENT_UNIQUE_ID"]
    .count()
)
#both are off (report and occ year), will try keeping only one record for duplicated occurences
#no dropping duplicate EVENT_UNIQUE_IDs, matches online dashboard


#date time consistency check
auto_theft_data_cleaning['Report_Year_Check'] = auto_theft_data_cleaning['REPORT_DATE'].dt.year
auto_theft_data_cleaning['Report_Month_Check'] = auto_theft_data_cleaning['REPORT_DATE'].dt.month_name()
auto_theft_data_cleaning['Report_Day_Check'] = auto_theft_data_cleaning['REPORT_DATE'].dt.day
auto_theft_data_cleaning['Report_DOW_Check'] = auto_theft_data_cleaning['REPORT_DATE'].dt.day_name()
auto_theft_data_cleaning['Report_Hour_Check'] = auto_theft_data_cleaning['REPORT_DATE'].dt.hour

auto_theft_data_cleaning['Report_DOW_Check'] = (
    auto_theft_data_cleaning['Report_DOW_Check']
    .str.strip()
    .str.lower()
)

auto_theft_data_cleaning['REPORT_DOW'] = (
    auto_theft_data_cleaning['REPORT_DOW']
    .str.strip()
    .str.lower()
)

#do they match?
mismatches = auto_theft_data_cleaning[
    (auto_theft_data_cleaning['Report_Year_Check'] != auto_theft_data_cleaning['REPORT_YEAR'])  |
    (auto_theft_data_cleaning['Report_Month_Check'] != auto_theft_data_cleaning['REPORT_MONTH'])  |
    (auto_theft_data_cleaning['Report_Day_Check'] != auto_theft_data_cleaning['REPORT_DAY'])  |
    (auto_theft_data_cleaning['Report_DOW_Check'] != auto_theft_data_cleaning['REPORT_DOW'])  |
    (auto_theft_data_cleaning['Report_Hour_Check'] != auto_theft_data_cleaning['REPORT_HOUR'])
]


##only report hour mismatches


print(auto_theft_data_cleaning["Report_Hour_Check"].min()) #4
print(auto_theft_data_cleaning["Report_Hour_Check"].max()) #5

#analysis shows time portion of the report_hour is not accurate




#date time consistency check
auto_theft_data_cleaning['Occ_Year_Check'] = auto_theft_data_cleaning['OCC_DATE'].dt.year
auto_theft_data_cleaning['Occ_Month_Check'] = auto_theft_data_cleaning['OCC_DATE'].dt.month_name()
auto_theft_data_cleaning['Occ_Day_Check'] = auto_theft_data_cleaning['OCC_DATE'].dt.day
auto_theft_data_cleaning['Occ_DOW_Check'] = auto_theft_data_cleaning['OCC_DATE'].dt.day_name()
auto_theft_data_cleaning['Occ_Hour_Check'] = auto_theft_data_cleaning['OCC_DATE'].dt.hour

auto_theft_data_cleaning['Occ_DOW_Check'] = (
    auto_theft_data_cleaning['Occ_DOW_Check']
    .str.strip()
    .str.lower()
)

auto_theft_data_cleaning['OCC_DOW'] = (
    auto_theft_data_cleaning['OCC_DOW']
    .str.strip()
    .str.lower()
)


#do they match?
mismatches_occ = auto_theft_data_cleaning[
    (auto_theft_data_cleaning['Occ_Year_Check'] != auto_theft_data_cleaning['OCC_YEAR'])  |
    (auto_theft_data_cleaning['Occ_Month_Check'] != auto_theft_data_cleaning['OCC_MONTH'])  |
    (auto_theft_data_cleaning['Occ_Day_Check'] != auto_theft_data_cleaning['OCC_DAY'])  |
    (auto_theft_data_cleaning['Occ_DOW_Check'] != auto_theft_data_cleaning['OCC_DOW']) |
    (auto_theft_data_cleaning['Occ_Hour_Check'] != auto_theft_data_cleaning['OCC_HOUR'])
]

##only occ hour mismatches


print(auto_theft_data_cleaning["Occ_Hour_Check"].min()) #4
print(auto_theft_data_cleaning["Occ_Hour_Check"].max()) #5

#analysis shows time portion of the occ_hour is not accurate

#captialize first day on DOW columns
cap_cols = ['REPORT_DOW', 'OCC_DOW']
auto_theft_data_cleaning[cap_cols] = auto_theft_data_cleaning[cap_cols].apply(lambda s: s.str.title())




##check lat/long values to see if in toronto city limits

#min_lat = 43.58
#max_lat = 43.86
print("Latitude range:",
      auto_theft_data_cleaning['LAT_WGS84'].min(), #0 - min shouldn't be 0
      auto_theft_data_cleaning['LAT_WGS84'].max()) #43.85 - under upper limit

#min_long = -79.64
#max_long = -79.12
print("Longitude range:",
      auto_theft_data_cleaning['LONG_WGS84'].min(), #-79.64 - above lower limit
      auto_theft_data_cleaning['LONG_WGS84'].max()) # 0 - max can't be 0


#investigate the 0 for lat and long
auto_theft_data_cleaning_zero_long_lat = auto_theft_data_cleaning[(auto_theft_data_cleaning["LAT_WGS84"] == 0) | 
                                                                  (auto_theft_data_cleaning["LONG_WGS84"] == 0)]

#both long and lat are 0?
num_zero_both = (
    (auto_theft_data_cleaning["LAT_WGS84"] == 0) &
    (auto_theft_data_cleaning["LONG_WGS84"] == 0)
).sum()

print(num_zero_both) #843
print(len(auto_theft_data_cleaning_zero_long_lat)) #843

#both columns are zero, invesitgate
auto_theft_data_cleaning_zero_long_lat_describe = auto_theft_data_cleaning_zero_long_lat.describe()
#numerical and date time variables have variability

#what about string columns
cols = ["REPORT_MONTH", "REPORT_DOW", "OCC_MONTH", "OCC_DOW", "DIVISION", "LOCATION_TYPE", "PREMISES_TYPE", "HOOD_158",
        "NEIGHBOURHOOD_158", "HOOD_140",
                "NEIGHBOURHOOD_140"]

value_count_dfs = {
    col: (
        auto_theft_data_cleaning_zero_long_lat[col]
        .value_counts(dropna=False)
        .reset_index()
        .rename(columns={"index": col, col: "count"})
    )
    for col in cols}


## "HOOD_158", "NEIGHBOURHOOD_158", "HOOD_140", "NEIGHBOURHOOD_140" all "NSA"
#what does nsa mean and represent?

#No Specified Address (NSA) includes any occurrences reported outside the City of Toronto limits or those that have no verified location
#source: https://open.toronto.ca/dataset/police-annual-statistical-report-reported-crimes/

#should leave in so it matches official dashboard


#Ready for joining to other databases



##################import weather data####################
#source: https://toronto.weatherstats.ca/download.html 

from pathlib import Path
import requests


# Environment Canada weather configuration
WEATHER_STATION_ID = 51459
WEATHER_STATION_NAME = "TORONTO INTL A"
WEATHER_YEARS = [2014, 2015, 2016,2017,2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025]

WEATHER_DATA_DIR = Path("data/raw/weather")
WEATHER_DATA_DIR.mkdir(parents=True, exist_ok=True)

WEATHER_DOWNLOAD_URL = (
    "https://climate.weather.gc.ca/climate_data/bulk_data_e.html"
)

# Use hourly data
timeframe = 1 


# Download the annual weather files
weather_files = []

for year in WEATHER_YEARS:
    for month in range(1, 13):
        file_path = WEATHER_DATA_DIR / f"toronto_weather_{year}_{month:02d}.csv"

        if file_path.exists():
            print(f"Already downloaded: {file_path.name}")
        else:
            print(f"Downloading weather data for {year}-{month:02d}")

            params = {
                "format": "csv",
                "stationID": WEATHER_STATION_ID,
                "Year": year,
                "Month": month,
                "Day": 1,
                "timeframe": 1,  
                "submit": "Download Data",
            }

            response = requests.get(
                WEATHER_DOWNLOAD_URL,
                params=params,
                timeout=120,
            )

            response.raise_for_status()
            file_path.write_bytes(response.content)

            print(f"Saved: {file_path}")

        weather_files.append(file_path)

#Verify the downloaded files
weather_file_inventory = pd.DataFrame(
    [
        {
            "file": file_path.name,
            "size_bytes": file_path.stat().st_size,
        }
        for file_path in weather_files
    ]
)


# Verify the beginning of each file
for file_path in weather_files:
    print(f"\n===== {file_path.name} =====")

    with open(file_path, "r", encoding="utf-8-sig") as file:
        print(file.readline().strip())


# Load the annual files
weather_frames = []
weather_row_counts = []

for file_path in weather_files:
    annual_weather = pd.read_csv(
        file_path,
        encoding="utf-8-sig",
        low_memory=False
    )

    annual_weather["source_file"] = file_path.name

    weather_frames.append(annual_weather)

    weather_row_counts.append(
        {
            "file": file_path.name,
            "rows": len(annual_weather),
            "columns": annual_weather.shape[1],
        }
    )

weather = pd.concat(
    weather_frames,
    ignore_index=True,
    sort=False
)

print(f"Loaded {len(weather):,} hourly weather records.")


# Preview and inspect
weather.head()
print(f"Rows: {weather.shape[0]:,}")
print(f"Columns: {weather.shape[1]:,}")
weather.info()


weather_row_counts = pd.DataFrame(weather_row_counts)



dfs = []

for file_path in weather_files:
    df = pd.read_csv(file_path)
    dfs.append(df)

weather_df = pd.concat(dfs, ignore_index=True)

print(weather_df.shape)
weather_df.head()


#works not make it be able to join with autotheft data
#create dow week, parse out the hour/just keep the hour, make month into a strong var
#should join on occurance dt

weather_df['string_month'] = pd.to_datetime(weather_df['Month'], format='%m').dt.month_name()


weather_df["Date/Time (LST)"] = pd.to_datetime(weather_df["Date/Time (LST)"])
weather_df["DOW"] = weather_df["Date/Time (LST)"].dt.day_name()

weather_df["Hour"] = weather_df["Time (LST)"].str.split(":").str[0].astype(int)


###join

auo_theft_and_weather = auto_theft_data_cleaning.merge(
    weather_df,
    how="left",
    left_on=["OCC_YEAR", "OCC_MONTH", "OCC_DAY", "OCC_DOW", "OCC_HOUR"],
    right_on=["Year", "string_month", "Day", "DOW", "Hour" ]
)


################################community profile
#source: https://open.toronto.ca/dataset/neighbourhoods/

# tor_neigh = pd.read_csv(r"C:\Users\russe\Downloads\Neighbourhoods - 4326.csv")
import geopandas as gpd

# Configuration
NEIGHBOURHOOD_DATA_DIR = Path("data/raw/neighbourhoods")
NEIGHBOURHOOD_DATA_DIR.mkdir(parents=True, exist_ok=True)

print(NEIGHBOURHOOD_DATA_DIR.exists())

NEIGHBOURHOOD_URL = (
    "https://gis.toronto.ca/arcgis/rest/services/"
    "cot_geospatial26/MapServer/71/query"
)

geojson_path = (
    NEIGHBOURHOOD_DATA_DIR
    / "toronto_neighbourhoods_158.geojson"
)

params = {
    "where": "1=1",
    "outFields": "*",
    "outSR": "4326",
    "f": "geojson",
}

if not geojson_path.exists():
    response = requests.get(
        NEIGHBOURHOOD_URL,
        params=params,
        timeout=120
    )
    response.raise_for_status()
    geojson_path.write_bytes(response.content)
    print("Downloaded neighbourhood file.")
else:
    print("Using existing file.")

# Load the GeoJSON
neighbourhoods = gpd.read_file(geojson_path)

print(f"Loaded {len(neighbourhoods)} neighbourhoods.")


# Inspect the geometry
print(neighbourhoods.geometry.geom_type.value_counts())








#########mapping 

event_counts = (
    auo_theft_and_weather
    .groupby("HOOD_158")
    .size()
    .reset_index(name="EventCount")
)


neighbourhoods["AREA_SHORT_CODE"] = neighbourhoods["AREA_SHORT_CODE"].astype(int)
event_counts["HOOD_158"] = (
    event_counts["HOOD_158"]
    .replace("NSA", "999")
    .astype(int)
)


print(neighbourhoods["AREA_SHORT_CODE"].dtype)
print(event_counts["HOOD_158"].dtype)
print(event_counts["HOOD_158"].head())

map_df = neighbourhoods.merge(
    event_counts,
    left_on="AREA_SHORT_CODE",
    right_on="HOOD_158",
    how="left"
)

map_df["EventCount"] = map_df["EventCount"].fillna(0)

ax = map_df.plot(
    column="EventCount",
    cmap="OrRd",
    legend=True,
    figsize=(12, 12),
    edgecolor="black",
    linewidth=0.2
)

ax.set_axis_off()



##individual
events_clean = auo_theft_and_weather[
    (auo_theft_and_weather["LONG_WGS84"] != 0)
    & (auo_theft_and_weather["LAT_WGS84"] != 0)
].copy()

events_gdf = gpd.GeoDataFrame(
    events_clean,
    geometry=gpd.points_from_xy(
        events_clean["LONG_WGS84"],
        events_clean["LAT_WGS84"]
    ),
    crs="EPSG:4326"
)



ax = neighbourhoods.plot(
    figsize=(12, 12),
    facecolor="white",
    edgecolor="black"
)

events_gdf.plot(
    ax=ax,
    color="red",
    markersize=2
)

print("Neighbourhood bounds:", neighbourhoods.total_bounds)
print("Event bounds:", events_gdf.total_bounds)




###start filtering my weather type -- need to create categories of weather
#cats - season [done], temp [done], hum[done] , rain[no hourly rain available], wind dir [done], wind spd, visibility, weather type [done]


auo_theft_and_weather["OCC_DATETIME"] = pd.to_datetime(auo_theft_and_weather["OCC_DATE"]).dt.normalize()
auo_theft_and_weather["OCC_DATETIME"] = pd.to_datetime(auo_theft_and_weather["OCC_DATETIME"]).dt.date



##### season flag
import numpy as np

auo_theft_and_weather["OCC_DATETIME"] = pd.to_datetime(
    auo_theft_and_weather["OCC_DATETIME"]
)

season_starts = {
    2014: {"Spring": "2014-03-20", "Summer": "2014-06-21", "Fall": "2014-09-22", "Winter": "2014-12-21"},
    2015: {"Spring": "2015-03-20", "Summer": "2015-06-21", "Fall": "2015-09-23", "Winter": "2015-12-21"},
    2016: {"Spring": "2016-03-20", "Summer": "2016-06-20", "Fall": "2016-09-22", "Winter": "2016-12-21"},
    2017: {"Spring": "2017-03-20", "Summer": "2017-06-21", "Fall": "2017-09-22", "Winter": "2017-12-21"},
    2018: {"Spring": "2018-03-20", "Summer": "2018-06-21", "Fall": "2018-09-22", "Winter": "2018-12-21"},
    2019: {"Spring": "2019-03-20", "Summer": "2019-06-21", "Fall": "2019-09-23", "Winter": "2019-12-21"},
    2020: {"Spring": "2020-03-19", "Summer": "2020-06-20", "Fall": "2020-09-22", "Winter": "2020-12-21"},
    2021: {"Spring": "2021-03-20", "Summer": "2021-06-20", "Fall": "2021-09-22", "Winter": "2021-12-21"},
    2022: {"Spring": "2022-03-20", "Summer": "2022-06-21", "Fall": "2022-09-22", "Winter": "2022-12-21"},
    2023: {"Spring": "2023-03-20", "Summer": "2023-06-21", "Fall": "2023-09-23", "Winter": "2023-12-21"},
    2024: {"Spring": "2024-03-19", "Summer": "2024-06-20", "Fall": "2024-09-22", "Winter": "2024-12-21"},
    2025: {"Spring": "2025-03-20", "Summer": "2025-06-20", "Fall": "2025-09-22", "Winter": "2025-12-21"},
}

def get_season(date):
    if pd.isna(date):
        return np.nan

    date = pd.Timestamp(date)
    year = date.year

    starts = season_starts.get(year)
    if starts is None:
        return np.nan

    spring = pd.Timestamp(starts["Spring"])
    summer = pd.Timestamp(starts["Summer"])
    fall = pd.Timestamp(starts["Fall"])
    winter = pd.Timestamp(starts["Winter"])

    if spring <= date < summer:
        return "Spring"
    elif summer <= date < fall:
        return "Summer"
    elif fall <= date < winter:
        return "Fall"
    else:
        return "Winter"

# Create Season column
auo_theft_and_weather["Season"] = auo_theft_and_weather["OCC_DATETIME"].apply(get_season)

# Convert to ordered categorical
auo_theft_and_weather["Season"] = pd.Categorical(
    auo_theft_and_weather["Season"],
    categories=["Winter", "Spring", "Summer", "Fall"],
    ordered=True
)

####temperature
temp_bins = [-np.inf, -15, -5, 5, 15, 25, np.inf]
temp_labels = [
    "Extreme Cold",
    "Cold",
    "Cool",
    "Mild",
    "Warm",
    "Hot"
]

auo_theft_and_weather["Temp_Category"] = pd.cut(
    auo_theft_and_weather["Temp (°C)"],
    bins=temp_bins,
    labels=temp_labels,
    ordered=True
)

	
###humdity
hum_bins = [0, 30, 50, 70, 90, 100]
hum_labels = [
    "Very Dry",
    "Dry",
    "Comfortable",
    "Humid",
    "Very Humid"
]

auo_theft_and_weather["Humidity_Category"] = pd.cut(
    auo_theft_and_weather["Rel Hum (%)"],
    bins=hum_bins,
    labels=hum_labels,
    include_lowest=True,
    ordered=True
)


###wind direction
def wind_dir_category(x):
    if pd.isna(x):
        return np.nan

    deg = x * 10  # convert from tens of degrees

    if deg >= 337.5 or deg < 22.5:
        return "N"
    elif deg < 67.5:
        return "NE"
    elif deg < 112.5:
        return "E"
    elif deg < 157.5:
        return "SE"
    elif deg < 202.5:
        return "S"
    elif deg < 247.5:
        return "SW"
    elif deg < 292.5:
        return "W"
    else:
        return "NW"

auo_theft_and_weather["Wind_Direction"] = (
    auo_theft_and_weather["Wind Dir (10s deg)"]
    .apply(wind_dir_category)
)

auo_theft_and_weather["Wind_Direction"] = pd.Categorical(
    auo_theft_and_weather["Wind_Direction"],
    categories=["N", "NE", "E", "SE", "S", "SW", "W", "NW"],
    ordered=False
)



##wind speed
wind_bins = [0, 10, 20, 40, 60, np.inf]
wind_labels = [
    "Light",
    "Moderate",
    "Strong",
    "Very Strong",
    "Extreme"
]

auo_theft_and_weather["WindSpeed_Category"] = pd.cut(
    auo_theft_and_weather["Wind Spd (km/h)"],
    bins=wind_bins,
    labels=wind_labels,
    include_lowest=True,
    ordered=True
)



#visibility
visibility_bins = [0, 5, 10, 20, np.inf]
visibility_labels = [
    "Low",
    "Moderate",
    "High",
    "Very High"
]

auo_theft_and_weather["Visibility_Category"] = pd.cut(
    auo_theft_and_weather["Visibility (km)"],
    bins=visibility_bins,
    labels=visibility_labels,
    include_lowest=True,
    ordered=True
)


###day/night
def time_period(hour):
    if 6 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 18:
        return "Afternoon"
    elif 18 <= hour < 24:
        return "Evening"
    else:
        return "Overnight"

auo_theft_and_weather["Time_Period"] = (
    auo_theft_and_weather["OCC_HOUR"]
    .apply(time_period)
)

auo_theft_and_weather["Time_Period"] = pd.Categorical(
    auo_theft_and_weather["Time_Period"],
    categories=["Overnight", "Morning", "Afternoon", "Evening"],
    ordered=True
)


##########################################################
###
###                 EDA
###
############################################################

#overall
total_thefts = len(auo_theft_and_weather)


#by year
thefts_by_year = (
    auo_theft_and_weather
    .groupby("REPORT_YEAR")
    .size()
    .reset_index(name="Thefts")
)

thefts_by_year.loc[len(thefts_by_year)] = [
"Total", thefts_by_year["Thefts"].sum()
]


#by month
thefts_by_month = (
    auo_theft_and_weather
    .groupby("REPORT_MONTH")
    .size()
    .reset_index(name="Thefts")
)

thefts_by_month.loc[len(thefts_by_month)] = [
"Total", thefts_by_month["Thefts"].sum()
]

#by year, and month
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

thefts_by_year_month = (
    auo_theft_and_weather
    .groupby(["REPORT_YEAR", "REPORT_MONTH"])
    .size()
    .reset_index(name="Thefts")
)

thefts_by_year_month["REPORT_MONTH"] = pd.Categorical(
    thefts_by_year_month["REPORT_MONTH"],
    categories=month_order,
    ordered=True
)

thefts_by_year_month = (
    thefts_by_year_month
    .sort_values(["REPORT_YEAR", "REPORT_MONTH"])
    .reset_index(drop=True)
)

total_row = pd.DataFrame({
    "REPORT_YEAR": ["Total"],
    "REPORT_MONTH": [""],
    "Thefts": [thefts_by_year_month["Thefts"].sum()]
})

thefts_by_year_month = pd.concat(
    [thefts_by_year_month, total_row],
    ignore_index=True
)


###viz
import seaborn as sns
import matplotlib.pyplot as plt

#line
thefts_by_year = (
    auo_theft_and_weather
    .groupby("REPORT_YEAR")
    .size()
    .reset_index(name="Thefts")
)

plt.figure(figsize=(10, 5))

sns.lineplot(
    data=thefts_by_year,
    x="REPORT_YEAR",
    y="Thefts",
    marker="o"
)

plt.title("Auto Thefts by Year")
plt.xlabel("Year")
plt.ylabel("Thefts")
plt.tight_layout()
plt.show()


#line_by_month
monthly_ts = (
    auo_theft_and_weather
    .groupby(["REPORT_YEAR", "REPORT_MONTH"])
    .size()
    .reset_index(name="Thefts")
)

monthly_ts["Date"] = pd.to_datetime(
    monthly_ts["REPORT_YEAR"].astype(str) + "-" +
    monthly_ts["REPORT_MONTH"].astype(str) + "-01"
)

monthly_ts = monthly_ts.sort_values("Date")

plt.figure(figsize=(15, 6))

sns.lineplot(
    data=monthly_ts,
    x="Date",
    y="Thefts",
    marker="o"
)

plt.title("Monthly Auto Thefts (2014-2025)")
plt.xlabel("Date")
plt.ylabel("Thefts")

plt.tight_layout()
plt.show()




#line
thefts_by_year = (
    auo_theft_and_weather
    .groupby("REPORT_YEAR")
    .size()
    .reset_index(name="Thefts")
)

plt.figure(figsize=(10, 5))

sns.lineplot(
    data=thefts_by_year,
    x="REPORT_YEAR",
    y="Thefts",
    marker="o"
)

plt.title("Auto Thefts by Year")
plt.xlabel("Year")
plt.ylabel("Thefts")
plt.tight_layout()
plt.show()


#looping the rest

def plot_heatmap(df, category_col, title=None):
    
    heatmap_data = (
        df.groupby(["REPORT_YEAR", category_col])
        .size()
        .reset_index(name="Thefts")
        .pivot(
            index="REPORT_YEAR",
            columns=category_col,
            values="Thefts"
        )
        .fillna(0)
    )

    plt.figure(figsize=(10, 6))

    sns.heatmap(
        heatmap_data,
        annot=True,
        fmt=".0f",
        cmap="YlOrRd",
        linewidths=0.5
    )

    plt.title(title or f"Thefts by Year and {category_col}")
    plt.xlabel(category_col)
    plt.ylabel("Year")
    plt.tight_layout()
    plt.show()


plot_heatmap(
    auo_theft_and_weather,
    "Season",
    "Auto Thefts by Year and Season"
)

plot_heatmap(
    auo_theft_and_weather,
    "Temp_Category",
    "Auto Thefts by Year and Temperature Category"
)

plot_heatmap(
    auo_theft_and_weather,
    "Humidity_Category",
    "Auto Thefts by Year and Humidity Category"
)

plot_heatmap(
    auo_theft_and_weather,
    "Wind_Direction",
    "Auto Thefts by Year and Wind Direction"
)

plot_heatmap(
    auo_theft_and_weather,
    "WindSpeed_Category",
    "Auto Thefts by Year and Wind Speed Category"
)

plot_heatmap(
    auo_theft_and_weather,
    "Visibility_Category",
    "Auto Thefts by Year and Visibility Category"
)

plot_heatmap(
    auo_theft_and_weather,
    "Time_Period",
    "Auto Thefts by Year and Time Period Category"
)



###pairplot
# pairplot_df = auo_theft_and_weather.drop(
# columns=["UCR_CODE", "UCR_EXT", "OFFENCE", "CSI_CATEGORY", "HOOD_158", "HOOD_140", "NEIGHBOURHOOD_140", "LONG_WGS84",
#          "LAT_WGS84", "x", "y", "Report_Year_Check", "Report_Month_Check", "Report_Day_Check", "Report_DOW_Check",
#          "Report_Hour_Check", "Report_Hour_Check", "Occ_Year_Check", "Occ_Month_Check", "Occ_Day_Check", "Occ_DOW_Check",
#          "Occ_Hour_Check", "Occ_Hour_Check", "Longitude (x)", "Latitude (y)", "Station Name", "Climate ID",
#          "Date/Time (LST)", "Flag", "Temp Flag", "Dew Point Temp Flag", "Rel Hum Flag", "Precip. Amount (mm)",
#          "Precip. Amount Flag", "Wind Dir Flag", "Wind Spd Flag", "Visibility Flag", "Stn Press Flag", "Hmdx Flag",
#          "Wind Chill Flag"])


daily_weather_thefts = (
    auo_theft_and_weather
    .groupby(auo_theft_and_weather["OCC_DATETIME"].dt.date)
    .agg(
        Thefts=("OCC_DATETIME", "size"),
        Temp=("Temp (°C)", "mean"),
        Humidity=("Rel Hum (%)", "mean"),
        WindSpeed=("Wind Spd (km/h)", "mean"),
        Visibility=("Visibility (km)", "mean")
    )
    .reset_index()
)

corr = daily_weather_thefts[
    ["Thefts", "Temp", "Humidity", "WindSpeed", "Visibility"]
].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Matrix: Daily Auto Thefts and Weather")
plt.show()


#scatt
plt.figure(figsize=(8, 6))

sns.regplot(
    data=daily_weather_thefts,
    x="Temp",
    y="Thefts",
    scatter_kws={"alpha": 0.3},
    line_kws={"color": "red"}
)

plt.title("Daily Auto Thefts vs Temperature")
plt.show()




sns.regplot(
    data=daily_weather_thefts,
    x="Humidity",
    y="Thefts",
    scatter_kws={"alpha": 0.3},
    line_kws={"color": "red"}
)

plt.title("Daily Auto Thefts vs Humidity")
plt.show()





sns.regplot(
    data=daily_weather_thefts,
    x="WindSpeed",
    y="Thefts",
    scatter_kws={"alpha": 0.3},
    line_kws={"color": "red"}
)

plt.title("Daily Auto Thefts vs Wind Speed")
plt.show()


#bar
temp_counts = (
    auo_theft_and_weather
    .groupby("Temp_Category")
    .size()
    .reset_index(name="Thefts")
)

sns.barplot(
    data=temp_counts,
    x="Temp_Category",
    y="Thefts"
)

plt.xticks(rotation=45)
plt.show()

```
